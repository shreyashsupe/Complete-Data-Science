```python
# Count the digit that divide a number

'''
Example 2:

Input: num = 121
Output: 2
Explanation: 121 is divisible by 1, but not 2. Since 1 occurs twice as a digit, we return 2.
'''

class Solution:
    def countDigits(self, num: int) -> int:
        temp = num
        ans = 0
        while temp > 0:
            r = temp % 10   # gives the last digit 
            if num % r == 0:
                ans += 1
            temp = temp//10  # chop of the last digit 

        
        return ans 
    
```
