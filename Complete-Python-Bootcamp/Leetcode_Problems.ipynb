```python
# Count the digit that divide a number

'''
Example 2:

Input: num = 121
Output: 2
Explanation: 121 is divisible by 1, but not 2. Since 1 occurs twice as a digit, we return 2.
'''

class Solution:
    def countDigits(self, num: int) -> int:
        temp = num
        ans = 0
        while temp > 0:
            r = temp % 10   # gives the last digit 
            if num % r == 0:
                ans += 1
            temp = temp//10  # chop of the last digit 

        
        return ans 
    
```

```python
# 9. Palindrome Number
# Given an integer x, return true if x is a palindrome, and false otherwise.

"""
Example 1:

Input: x = 121
Output: true
Explanation: 121 reads as 121 from left to right and from right to left.
"""

class solution:
    def isPalindrome(self, x):
        temp = x 
        reverse = 0

        while temp>0:
            r = temp%10          # get the last digit of the temp
            temp = temp // 10    # remove the last digit of the temp
            reverse = reverse * 10 + r

        return reverse == x 


# Create and object of the class soution 
sol = solution()       
sol.isPalindrome(121)
```

```python
# 9. Palindrome Number
# Given an integer x, return true if x is a palindrome, and false otherwise.

"""
Example 1:

Input: x = 121
Output: true
Explanation: 121 reads as 121 from left to right and from right to left.
"""

class solution:
    def isPalindrome(self, x):
        return str(x) == str(x)[::-1]

sol = solution()       
sol.isPalindrome(121)
```

```python
# 412. Fizz Buzz
"""
Given an integer n, return a string array answer (1-indexed) where:

answer[i] == "FizzBuzz" if i is divisible by 3 and 5.
answer[i] == "Fizz" if i is divisible by 3.
answer[i] == "Buzz" if i is divisible by 5.
answer[i] == i (as a string) if none of the above conditions are true.

Example :

Input: n = 15
Output: ["1","2","Fizz","4","Buzz","Fizz","7","8","Fizz","Buzz","11","Fizz","13","14","FizzBuzz"]
"""

class Solution:
    def fizzBuzz(self, n: int):
        ans = []
        for i in range(1,n+1):
            if i%3==0 and i%5==0:
                ans.append("FizzBuzz")
            elif i%3==0:
                ans.append("Fizz")
            elif i%5==0:
                ans.append("Buzz")
            else:
                ans.append(str(i))
            
        return ans 
    
```
