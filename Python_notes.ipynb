#### Time Complexity 

Time complexity describes how the execution time of an algorithm grows as the input size increases. It essentially qualifies the number of operations an algorithm as a function of the input data size.

We can tell the relation as the input size grows, the time taken to execute based on the algorithm complexity.

We can find the time complexity of the program, because time complexity is exactly propotional to the number of operation that are being done in a program.

Time complexity basically means number of operations.

Time complexity tells us how the number of operations grows as the input size (n) increases.


#### Asymptotic notations:

- Notations to donate the time complexity of the program.

Best case () - Min number fo operation that will take place 

Average case (0) - The maximum number of operations the program could take for an input of size n

Worst case (o) - expected operations over random inputs.

```python
n = 10 
x = 3
for i in range(n):
    if i == x:
        break 
    print(i, end=" ")

"""
Best case - omega(1)
Worst case - O(n)
Average case - 0(n/2)
"""
# Time complexity depends upon the number of operations in the program, in the above program the for loop is being executed for 3 time but for different values of x the operation will be different. Suppose x = 0 then the number of operation will happens are 1 that why the best case here is that min 1 operation will happens, and then the worst case is n number of operations will take place and for average case we can say the average n/2.
```

#### Types of time complexities





```python
# 1. Linear time complexities O(n)

"""
The time taken grows in direct proportion to the size of the input.
If input size doubles, the operations also double.
If input size becomes 10× bigger, the operations also become about 10× bigger.

Example:
Imagine you’re checking names in a class list one by one.
If the class has 10 students, you may take ~10 checks.
If the class has 100 students, you’ll take ~100 checks.

So → time grows linearly with input size.
"""
"""
If arr has 5 elements → 5 print operations.
If arr has 1000 elements → 1000 print operations.
Time taken grows linearly with n → O(n).
"""

def print_elements(arr):
    for x in arr:
        print(x)

```

```python
# Constant time complexity O(1)
"""
The number of operations does not depend in input size. No matter how big the input is, thw work is the same.
Input = 10 → still 1 step
Input = 10,000 → still 1 step
"""

# EXample 1 
# Accessing the elements in a list by index

arr = [10, 20, 30, 40, 50]
print(arr[3])  # Always O(1)

# Doesn’t matter if arr has 5 elements or 5 million — accessing arr[3] takes the same time.

# Example 2
# Checking if number is even 

n = 273
print(n%2==0) # always O(1)

# Just one calculation, independent of input size.

# example 3

for i in range(10000):
    print(i, end=" ")

```

```python
# Quadratic Time Complexity O(n²)
"""
The number of operations grows like the square of the input size
If input size = n, work ≈ n × n.
If input doubles, work becomes 4× bigger.
If input triples, work becomes 9× bigger.

Example:
Imagine a class with n students.
Each student shakes hands with every other student.
That’s about n × n handshakes → quadratic growth.
"""

# Example 1 Nested loop
def print_pairs(arr):
    n = len(arr)
    for i in range(n):
        for j in range(n):
            print(arr[i], arr[j])

# Outer loop runs n times. Inner loop runs n times for each outer loop. Total ≈ n × n = n² operations.
```

```python
# Logarithmic time complexity O(log n)
"""
Logarithmic means → the number of operations grows very slowly as input size increases.
Each step reduces the problem size by a fraction (usually half).
If input size doubles, the work increases by just 1 extra step.

Example:
Think of finding a word in a dictionary:
You don’t read page by page.
Instead, you open the middle, then decide left or right.
Each time you halve the search space.
So finding 1 word in 1,000 pages takes about 10 steps, not 1,000.

Logarithmic time complexity (O(log n)) happens when the problem size gets cut down by a fraction (often half) in each step. It grows very slowly and is super efficient.
"""

# Example 1:
# Keep dividing a number by 2 untill it becomes 1
def divide(n):
    steps = 0
    while n>1:
        n = n // 2
        steps = steps + 1
    return steps

# n = 16 → 16 → 8 → 4 → 2 → 1 → 4 steps
# n = 32 → 32 → 16 → 8 → 4 → 2 → 1 → 5 steps
# Each step divides by 2, so the steps ≈ log₂(n).
```

#### Space Complexity 

Refers to the amount of memory an algorithm used as a function of the input size. Its a measure how much memory an algorithm requires to to execute and solve a problem.

It tells you how much extra memory (RAM) a program needs while running, depending on the input size.

Example:

O(1) → uses fixed memory (just a few variables).

O(n) → memory grows with input size (like storing an array of n items).

O(n²) → memory grows with the square of input (like storing a 2D grid of n×n).



#### Recursion

The process in which function calls itself directly or in directly is called recursion and the corresponding function is called recursive function 

Loops are memory-efficient and faster, best for simple linear tasks.
Recursion is elegant for naturally recursive problems (trees, DFS) but uses more memory and can be slower.

```python
# Print 1 to n  

# using while loop
n = 5  
i = 1
while i<=5:
    print(i, end=" ")
    i = i+1
```

```python
# Print 1 to n using the recursion

def printNumber(i,n):
    #base case (stops the recursion to prevent infinite calls.)
    if i > n:
        return 
    #recursive case (the part where the function calls itself to solve a smaller problem.)
    print(i, end=" ")
    printNumber(i+1, n)


# call the function 
printNumber(1,5)

```

```python
# Factorial of a number

def  factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
    

# Call the function
print(factorial(4))
```

```python
# Recursive stack 
"""
Every time a function calls itself, Python stores the current function’s state (variables, execution point) on the call stack.
When a function reaches its base case, it starts returning, and the stack unwinds — meaning each stored call is completed in reverse order.
"""

"""
How the Recursive Stack Works

factorial(3) → needs 3 * factorial(2) → pause 3, store state on stack
factorial(2) → needs 2 * factorial(1) → pause 2, store state on stack
factorial(1) → needs 1 * factorial(0) → pause 1, store state on stack
factorial(0) → returns 1 → base case reached, stack starts unwinding

Now stack unwinds:

factorial(1) → 1 * 1 = 1
factorial(2) → 2 * 1 = 2
factorial(3) → 3 * 2 = 6
"""
```
